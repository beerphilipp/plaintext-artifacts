# Evaluation Script: General Dataset Stats

This script produces results for Section 4.1 Dataset

### Connect to the databases

```python
import subprocess
import sqlite3

SQLITE_DB_PATH = "/data/gplay.sqlite"
APK_DIR_SSH_USER = "dl"
APK_DIR_SSH_HOST = "download.st1.secpriv.wien"
APK_DIR_SSH_KEY = "~/.ssh/webview_key"
APK_DIR_SSH_FOLDER = "."

conn = sqlite3.connect(SQLITE_DB_PATH)
cursor = conn.cursor()

def print_latex_macro(name: str, value: str):
    print(f"\\newcommand{{\\{name}}}{{{value}}}")
```

### Retrieve the total downloaded apps

```python
cmd = (
      f"rsync -e 'ssh -p 22111 -i {APK_DIR_SSH_KEY} -o StrictHostKeyChecking=no -o UserKnownHostsFile=/dev/null' -zdn --out-format='%n' {APK_DIR_SSH_USER}@{APK_DIR_SSH_HOST}: . | wc -l"
  )
total_downloaded_apps = int(subprocess.check_output(cmd, shell=True).strip())
print_latex_macro("totalDownloadedApps", f"{total_downloaded_apps:,}")
```

### Retrieve stats on the dataset

```python
cursor.execute("SELECT COUNT(*) FROM package_name;")
total_apps = cursor.fetchone()[0]
print_latex_macro("totalPackageNames", f"{total_apps:,}")

# Count the amount of packages that have metadata
cursor.execute("SELECT COUNT(*) FROM info WHERE success = 1;")
total_apps_with_metadata = cursor.fetchone()[0]
print_latex_macro("totalPackageNamesWithMetadata", f"{total_apps_with_metadata:,}")

# Count the amount of packages that are free
cursor.execute("SELECT COUNT(*) FROM info WHERE free = 1;")
total_apps_free = cursor.fetchone()[0]
print_latex_macro("totalPackageNamesFree", f"{total_apps_free:,}")

# Count the amount of packages that are free and have at least 100,000 installs
cursor.execute("SELECT COUNT(*) FROM info WHERE free = 1 AND min_installs >= 100000;")
total_apps_free_100k = cursor.fetchone()[0]
print_latex_macro("totalPackageNamesFreeMinHundredK", f"{total_apps_free_100k:,}")
```
