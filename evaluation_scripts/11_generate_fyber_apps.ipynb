# Evaluation Script: Fyber

This script produces results for Section 6.2 App Interference

### Connect to the MongoDB

```python
import pymongo

client = pymongo.MongoClient("mongodb://localhost:27017/")
db = client["webview"]
network_logs_collection = db["network_logs"]

def print_latex_macro(name: str, value: str):
    print(f"\\newcommand{{\\{name}}}{{{value}}}")
```

#### Retrieve those apps where we detect HTTP traffic from Fyber

Produces results for "Fyber SDK"

```python
package_names = network_logs_collection.distinct("package_name", {"initiator": "http://wv.inner-active.mobi"})
print_latex_macro("appsFyberHTTP", f"{len(package_names):,}")
```
